## 1. My lane (or freestyle) and why

I selected the Refresh / Content Opportunity Scoring lane.

My goal is to identify which pages should be reviewed first when editorial time is limited. This lane can help a content team decide whether a page should be refreshed, expanded, protected, or monitored.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.import pandas as pd
import pandas as pd

data_url = "https://raw.githubusercontent.com/shaestasaleem/flyrank-machine-learning-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_url)

print("Dataset loaded successfully")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

## 2. The question: decision, action, cost of a wrong call

My research question is: Which content pages should be reviewed first when editorial resources are limited, based on observed search visibility, content age, ranking position, clicks, impressions, and engagement signals?

The unit of analysis is one content page.

The output will be a ranked list of pages that need attention.

The action will be to refresh, expand, protect, or monitor a page.

A wrong recommendation may waste editorial time or cause an important page to be ignored.

Data and machine learning may help because several signals need to be considered together.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("Unique pages:", df["content_id"].nunique())
print("Unique clients:", df["client_id"].nunique())
print("Unit of analysis: one content page")
```

## 3. Quick look at the data (2-3 real numbers)


The starter dataset contains 30,000 content pages from 32 clients.

Among these pages, 22,006 have at least 100 impressions, which shows that a large number of pages have enough search visibility to support meaningful analysis.

The dataset also contains 6,360 pages that are more than one year old. These observed numbers suggest that there may be useful opportunities to identify and prioritize pages for content review or refresh.

These numbers support my choice of the Refresh / Content Opportunity Scoring lane for further investigation.

```python

total_pages = len(df)
unique_clients = df["client_id"].nunique()
pages_with_demand = (df["impressions_90d"] >= 100).sum()
older_pages = (df["content_age_days"] >= 365).sum()

print("Total pages:", total_pages)
print("Unique clients:", unique_clients)
print("Pages with at least 100 impressions:", pages_with_demand)
print("Pages older than one year:", older_pages)
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').* This work can identify observed and directional patterns in the dataset. It can provide decision-support by helping content editors decide which pages should be reviewed first.

It cannot prove that refreshing a page will cause its Google ranking to improve. It also cannot predict Google's algorithm or guarantee better search performance.

The final recommendations should support human decisions rather than make automatic decisions.

```python
required_columns = [
    "content_id",
    "client_id",
    "impressions_90d",
    "content_age_days"
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

print("Missing required columns:", missing_columns)
print("This analysis provides observed and directional decision-support.")
print("It does not prove causal improvement or predict Google's algorithm.")
```
